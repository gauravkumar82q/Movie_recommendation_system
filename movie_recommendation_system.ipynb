```python
# =========================================
# 1️⃣ Import required libraries
# =========================================
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# =========================================
# 2️⃣ Load your dataset from sample_data
# =========================================
df = pd.read_csv("/content/sample_data/movies.csv")

# Show first 5 rows to confirm
print("Dataset loaded successfully!")
display(df.head())

# =========================================
# 3️⃣ Check and fill missing values
# =========================================
if 'genres' not in df.columns:
    raise ValueError("Your CSV must contain a 'genres' column.")

df['genres'] = df['genres'].fillna('')
df['title'] = df['title'].fillna('')

# =========================================
# 4️⃣ Convert genres text → vectors
# =========================================
cv = CountVectorizer()
genre_vectors = cv.fit_transform(df['genres'])

# =========================================
# 5️⃣ Compute cosine similarity
# =========================================
similarity = cosine_similarity(genre_vectors)

# =========================================
# 6️⃣ Improved Recommendation Function
#     ✔ Works with partial search
#     ✔ Shows closest matching title
# =========================================
def recommend(movie_name):
    movie_name = movie_name.lower()

    # Partial matching
    matches = df[df['title'].str.lower().str.contains(movie_name)]

    if matches.empty:
        return {"error": "Movie not found. Try typing part of the title."}

    # Use the first matched movie
    index = matches.index[0]
    searched_movie = df.iloc[index].title

    # Get similarity scores
    distances = list(similarity[index])
    movie_list = sorted(
        list(enumerate(distances)),
        reverse=True,
        key=lambda x: x[1]
    )[1:11]  # top 10 recommendations

    # Recommended movie titles
    recommended_titles = [df.iloc[i[0]].title for i in movie_list]

    return {
        "searched_movie": searched_movie,
        "recommendations": recommended_titles
    }

# =========================================
# 7️⃣ TESTING THE SYSTEM
# =========================================
print(recommend("toy"))      # try with partial name
print(recommend("dark"))     # another test
print(recommend("king"))     # another test
```
